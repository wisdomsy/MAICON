### 정리

```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("health_summary.csv")

# 1. 전체 병사에 대한 기술통계량 계산
overall_stats = df[["temperature", "pulse", "weight"]].agg(['mean', 'median', 'std'])

# 2. 부대별 평균 및 표준편차 계산
grouped_stats = df.groupby("unit")[["temperature", "pulse", "weight"]].agg(['mean', 'std'])

# 3. 저장 포맷 정리: 전체 + 부대별
# 전체 통계량은 행 인덱스를 붙이고, 부대별은 컬럼 멀티인덱스를 문자열로
overall_stats.index.name = "stat"
overall_stats_reset = overall_stats.reset_index()

grouped_stats.columns = ['_'.join(col).strip() for col in grouped_stats.columns.values]
grouped_stats_reset = grouped_stats.reset_index()

# 병합 후 저장
final_stats = pd.concat([overall_stats_reset, grouped_stats_reset], axis=1)

output_path = "health_stats.csv"
final_stats.to_csv(output_path, index=False)
```

### 정리

```python
import pandas as pd

# 데이터 로드
df_eq = pd.read_csv("equipment_usage.csv")

# 1. 전체 장비에 대한 사용 시간 통계
overall_stats_eq = df_eq["usage_hours"].agg(['mean', 'std', 'min', 'max', 'median']).to_frame().T
overall_stats_eq.index = ['전체']

# 2. 장비 유형별 통계량
type_stats_eq = df_eq.groupby('type')["usage_hours"].agg(['mean', 'std', 'min', 'max', 'median'])

# 3. 전체 + 유형별 통계 합치기 → 저장
final_stats_eq = pd.concat([overall_stats_eq, type_stats_eq])

# 저장
equipment_stats_path = "equipment_stats.csv"
final_stats_eq.to_csv(equipment_stats_path)
```

### 정리

```python
import pandas as pd

# 데이터 로드
df_meal = pd.read_csv("meal_survey.csv")

# 1. 전체 평균, 중앙값, 최빈값 계산
score_cols = ["taste_score", "quantity_score", "cleanliness_score"]

mean_scores = df_meal[score_cols].mean().to_frame(name="mean").T
median_scores = df_meal[score_cols].median().to_frame(name="median").T
mode_scores = df_meal[score_cols].mode().iloc[0].to_frame(name="mode").T

overall_summary = pd.concat([mean_scores, median_scores, mode_scores])
overall_summary.index.name = "stat"

# 2. 부대별 평균 및 표준편차 계산
unit_stats = df_meal.groupby("unit")[score_cols].agg(["mean", "std"])

# 멀티인덱스 컬럼 평탄화
unit_stats.columns = ['_'.join(col).strip() for col in unit_stats.columns.values]
unit_stats.reset_index(inplace=True)

# 3. 저장
survey_stats_path = "survey_unit_mean.csv"
unit_stats.to_csv(survey_stats_path, index=False)

overall_summary_path = "survey_overall_stats.csv"
overall_summary.to_csv(overall_summary_path)
```
